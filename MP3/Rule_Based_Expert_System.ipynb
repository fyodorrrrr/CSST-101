```python
# 📌 Simple Hobby Recommender (Personality + Interest)

def get_valid_input(prompt, valid_options):
    """
    Prompt the user until they enter a valid response.
    """
    while True:
        user_input = input(prompt).strip().lower()
        if user_input in valid_options:
            return user_input
        else:
            print(f"⚠️ Invalid input. Please choose from: {', '.join(valid_options)}")

# Step 1: Welcome and Inputs
print("🎯 Welcome to the Hobby Recommender!")

personality = get_valid_input(
    "Are you more of an introvert or extrovert? ",
    ["introvert", "extrovert"]
)

interest = get_valid_input(
    "What type of activities do you enjoy? (creative/physical/tech): ",
    ["creative", "physical", "tech"]
)

# Step 2: Inference Rules
print("\n✅ Recommended Hobby:")

if personality == "introvert":
    if interest == "creative":
        print("📌 Try journaling, drawing, or solo painting.")
    elif interest == "physical":
        print("📌 Try yoga, hiking, or cycling.")
    elif interest == "tech":
        print("📌 Try coding, online puzzle games, or Arduino kits.")

elif personality == "extrovert":
    if interest == "creative":
        print("📌 Try theater, group art classes, or spoken word.")
    elif interest == "physical":
        print("📌 Try team sports, dance classes, or martial arts.")
    elif interest == "tech":
        print("📌 Try hackathons, coding clubs, or game dev meetups.")

# Step 3: End
print("\nThanks for using the Hobby Recommender! 🎉")
```
